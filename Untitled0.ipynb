```python
df = pd.read_csv('fraud_oracle.csv')
#Display first 20 list
print(df.head(20))
```

```python
# The inspection and preparation stage where the structure of the data is checked.
# This is to understand the data and check its quality.
print(df.info())

#Check missing values
print(df.isnull().sum())

#Check the number of rows and columns
print(df.shape)

#Check whether there are duplicate records
print(df.duplicated().sum())
```

```python
# The actual analysis
#Computing or calculating the total number of PolicyTypes.

total_policytypes = df['PolicyType'].nunique()
print(f"Total number of policytypes: {total_policytypes}")
```

```python

age_policyholders = df["AgeOfPolicyHolder"].describe()
print(age_policyholders)
```

```python
#Checking Policy Types in the column
print(df["PolicyType"].value_counts())
```

```python
#Creating a bar chart
policy_counts = df["PolicyType"].value_counts()
plt.figure(figsize = (10,5))

sns.barplot(
    x=policy_counts.index,
    y=policy_counts.values,
)

plt.title("Distribution of Policy Type")
plt.xlabel("Policy Types")
plt.ylabel("Number of PolicyHolders")
plt.xticks(rotation=45)

plt.show()
```

```python
#Ccreating pie chart
policy_counts = df["Make"].value_counts()
plt.figure(figsize = (10,5))

plt.pie(
    policy_counts.values,
    labels=policy_counts.index,
    autopct="%1.1f%%",
)

plt.title("Distribution of Make")
plt.show()
```

```python
#Creating pivot tables for PolicyType, Month, Marriage Status and Deductable

# Average deductabiles associated with each PolicyType
pivot = pd.pivot_table(
    df,
    index="PolicyType",
    columns="Month",
    values="Deductible",
    aggfunc="mean",
)

print(pivot)
```

```python
#Calculating  total deductabiles by policy type
pivot = pd.pivot_table(
    df,
    index="PolicyType",
    values="Deductible",
    aggfunc="sum",
)

print(pivot)
```

```python
#Comparirng PolicyType by Marriage Status
pivot = pd.pivot_table(
    df,
    index="PolicyType",
    columns="MaritalStatus",
    values="Deductible",
    aggfunc="count",
    fill_value  = 0
)

print(pivot)
```

```python
#8. Visualising the  Pivot Table
pivot = pd.pivot_table(
    df,
    index="PolicyType",
    columns="MaritalStatus",
    values="Deductible",
    aggfunc="count",
    fill_value  = 0
)
pivot.plot(
    kind = "bar" ,
    figsize=(10, 6),
    legend = False,
    title = "Total deductables by PolicyType and Marital Status"
)
plt.xlabel("PolicyType")
plt.ylabel("Total Deductibles")
plt.tight_layout()
plt.show()

```

```python
#Indicates whether fraud was found.
print(df["FraudFound_P"].value_counts())


#Calculating the percentage of records marked as fraudulent.
fraud_counts = df["FraudFound_P"].value_counts()
total_records = fraud_counts.sum()
fraud_percentage = (fraud_counts[1] / total_records) * 100
print(f"Percentage of records marked as fraudulent: {fraud_percentage:.2f}%")

#Visualising the result.
sns.countplot(data=df, x="FraudFound_P")
plt.title("Distribution of Fraud Records")
plt.xlabel("Fraud Found")
plt.ylabel("Count")
plt.show()
```

```python
#Next, investigating fraud by policy type.
fraud_by_policytype = pd.crosstab(df["PolicyType"], df["FraudFound_P"], normalize="index")*100
print(fraud_by_policytype)


#Visualising the fraud percentages by PolicyType.
fraud_by_policytype.plot(
    kind="bar",
    figsize=(10, 6),
    legend=False,
    title="Fraud Percentage by PolicyType"
)
plt.xlabel("PolicyType")
plt.ylabel("Fraud Percentage")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
#Analysing fraud by marital status
#Use the MaritalStatus column to examine whether fraud rates differ across marital-status categories.
fraud_by_maritalstatus = pd.crosstab(df["MaritalStatus"], df["FraudFound_P"], normalize="index")*100
print(fraud_by_maritalstatus)

#Visualising the fraud percentages by MaritalStatus.
fraud_by_maritalstatus.plot(
    kind="bar",
    figsize=(10, 6),
    legend=False,
    title="Fraud Percentage by Marital Status"
)
plt.xlabel("Marital Status")
plt.ylabel("Fraud Percentage")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Investigating Fraud by accident Area
fraud_accident_area = pd.crosstab(df["AccidentArea"], df["FraudFound_P"], normalize="index")*100
print(fraud_accident_area)

#Visualizing the fraud Percentage by Accident Area
fraud_accident_area.plot(
    kind="bar",
    figsize=(10, 6),
    legend=False,
    title="Fraud Percentage by Accident Area"
)
plt.xlabel("Accident Area")
plt.ylabel("Fraud Percentage")
```

```python
#Investigating whether a police report was filed:
fraud_by_report= pd.crosstab(df["PoliceReportFiled"], df["FraudFound_P"], normalize="index")*100
print(fraud_by_report)

```

```python
#6.Testing the relationships between variables using statistical tests
#Visual comparisons can reveal patterns, but
#statistical tests help assess whether an observed relationship is likely to be meaningful.
#Using the chi-square test

# We want to know whether fraud associated with policy type?
from scipy.stats import chi2_contingency

table = pd.crosstab(df["PolicyType"], df["FraudFound_P"])
chi2, p_value, dof, expected = chi2_contingency(table)
print(f"Chi-square statistic: {chi2}")
print(f"P-value: {p_value}")

if p_value < 0.05:
    print("There is a significant association between policy type and fraud.")
else:
    print("There is no significant association between policy type and fraud.")
```
